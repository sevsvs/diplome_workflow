```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\boris\Documents\GitHub\diplome_workflow\datasets\final_dataset\structures.csv")
df
```

```python
crystal_system_counts = df['crystal_system'].value_counts()
plt.figure(figsize=(10, 6))
ax = crystal_system_counts.plot(kind='bar', color='skyblue', edgecolor='black')
ax.bar_label(ax.containers[0], padding=3, fontsize=11)
plt.xlabel('Сингония', fontsize=12)
plt.ylabel('Количество', fontsize=12)
plt.xticks(rotation=0)

plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
